```python
numeric_cols = df.select_dtypes(include=[np.number]).columns
corr_matrix = df[numeric_cols].corr()

high_corr_pairs = []
for i in range(len(corr_matrix.columns)):
    for j in range(i+1, len(corr_matrix.columns)):
        if abs(corr_matrix.iloc[i, j]) > 0.95:
            high_corr_pairs.append((corr_matrix.columns[i], 
                                    corr_matrix.columns[j], 
                                    corr_matrix.iloc[i, j]))

print(f"High correlation pairs (>0.95): {len(high_corr_pairs)}")
for feat1, feat2, corr in high_corr_pairs[:30]:
    print(f"  {feat1} <-> {feat2}: {corr:.3f}")
```

```python
def z_score(col):
    z= ((df[col]-df[col].mean())/df[col].std()).abs()
    outlier = (z>3).sum()
    return outlier

def IQR_method(col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outlier = ((df[col]<lower) | (df[col]>upper)).sum()
    return outlier

num_cols = [x for x in df.columns if df[x].dtype in ["int64", "float64"]]
cat_cols = [x for x in df.columns if df[x].dtype == "object"]

for col in num_cols:
    z = z_score(col)
    IQR = IQR_method(col)
    print(f"Column name: {col} - Z score: {z}, IQR: {IQR}")
```

```python

targets = {
    'success_label': {
        'type': 'classification',
        'threshold': 0.85,  
        'description': 'Will product succeed?'
    },
    'still_operating': {
        'type': 'classification',
        'threshold': 0.85,
        'description': 'Will product survive?'
    },
    'estimated_break_even_months': {
        'type': 'regression',
        'threshold': 0.80,
        'description': 'When will I break even?'
    },
    'revenue_estimated': {
        'type': 'regression',
        'threshold': 0.90,  
        'description': 'What revenue will I make?'
    },
    'risk_score': {
        'type': 'regression',
        'threshold': 0.85,
        'description': 'What is my risk level?'
    },
    'market_fit_score': {
        'type': 'regression',
        'threshold': 0.80,
        'description': 'How good is my market fit?'
    },
    'failure_reason_cat': {
        'type': 'classification',
        'threshold': 0.75,
        'description': 'What will cause failure?'
    }
}

ALWAYS_LEAKAGE = {
    'success_label': [
        'success_label', 'still_operating', 'is_dead', 'survival_months',
        'shutdown_date', 'exit_status', 'success_reason', 'success_reason_cat',
        'failure_reason', 'failure_reason_cat',
        'survival_score', 'growth_score', 'market_fit_score', 'risk_score',
        'revenue_monthly', 'revenue_monthly_filled', 'revenue_estimated',
        'monthly_profit_est', 'is_profitable',
        'capital_efficiency', 'revenue_to_burn_ratio'
    ],
    'still_operating': [
        'still_operating', 'is_dead', 'shutdown_date', 'survival_months',
        'exit_status', 'survival_score'
    ],
    'estimated_break_even_months': [
        'estimated_break_even_months', 'base_break_even_months',
        'monthly_profit_est', 'is_profitable'
    ],
    'revenue_estimated': [
        'revenue_monthly', 'revenue_monthly_filled', 'revenue_estimated',
        'monthly_profit_est', 'is_profitable',
        'capital_efficiency', 'revenue_to_burn_ratio'
    ],
    'risk_score': [
        'risk_score', 'survival_score', 'growth_score', 'market_fit_score'
    ],
    'market_fit_score': [
        'market_fit_score', 'survival_score', 'growth_score', 'risk_score'
    ],
    'failure_reason_cat': [
        'failure_reason_cat', 'failure_reason', 'success_reason_cat',
        'success_reason', 'risk_score'
    ]
}


leakage_report = {}

for target, config in targets.items():
    print(f"\n{'='*70}")
    print(f"🎯 TARGET: {target}")
    print(f"   Description: {config['description']}")
    print(f"   Type: {config['type']}")
    print(f"   Correlation threshold: {config['threshold']}")
    print(f"{'='*70}")
    
    # Initialize leakage list for this target
    leakage_features = set(ALWAYS_LEAKAGE.get(target, []))
    
    # 1. PREPARE THE DATA
    temp_df = df.copy()
    
    # Check if target exists
    if target not in temp_df.columns:
        print(f"⚠️ WARNING: {target} not found in dataset. Skipping.")
        continue
    
    # SPECIAL HANDLING: Encode categorical targets temporarily
    target_values = temp_df[target].copy()
    if temp_df[target].dtype == 'object':
        print(f"ℹ️ Encoding categorical target '{target}' for correlation check...")
        temp_df[target] = temp_df[target].astype('category').cat.codes
    
    # SPECIAL HANDLING: Filter valid rows only
    if target in ['success_label']:
        # Only use rows with success_label = 0 or 1 (exclude -1 = unknown)
        valid_mask = temp_df[target].isin([0, 1])
        temp_df = temp_df[valid_mask]
        print(f"ℹ️ Filtered to {len(temp_df)} valid rows (excluding unknown)")
    
    # 2. SELECT NUMERIC FEATURES ONLY
    numeric_cols = temp_df.select_dtypes(include=[np.number]).columns.tolist()
    
    # Remove target itself and already-known leakage
    numeric_cols = [col for col in numeric_cols if col not in leakage_features]
    
    if len(numeric_cols) == 0:
        print("⚠️ No numeric features to check!")
        continue
    
    # 3. CALCULATE CORRELATIONS
    try:
        correlations = temp_df[numeric_cols].corrwith(temp_df[target]).abs()
        correlations = correlations.sort_values(ascending=False)
        
        # Remove NaN correlations (constant columns)
        correlations = correlations.dropna()
        
    except Exception as e:
        print(f"❌ Error calculating correlations: {e}")
        continue
    
    # 4. DETECT HIGH CORRELATIONS (POTENTIAL LEAKAGE)
    threshold = config['threshold']
    high_corr = correlations[correlations > threshold]
    
    if len(high_corr) > 0:
        print(f"\n🚨 FOUND {len(high_corr)} POTENTIAL LEAKS (correlation > {threshold}):")
        print(high_corr.to_string())
        
        # Add to leakage set
        leakage_features.update(high_corr.index.tolist())
    else:
        print(f"✅ No high correlations found (all < {threshold})")
    
    # 5. STORE FINAL LEAKAGE LIST
    leakage_report[target] = sorted(list(leakage_features))
    
    print(f"\n📋 TOTAL LEAKAGE FEATURES FOR '{target}': {len(leakage_features)}")


print("\n" + "="*70)
print("📊 LEAKAGE DETECTION COMPLETE - FINAL REPORT")
print("="*70)

for target, leakage_list in leakage_report.items():
    print(f"\n🎯 {target.upper()}")
    print(f"   Leakage features to DROP: {len(leakage_list)}")
    print(f"   Features: {', '.join(leakage_list[:10])}{'...' if len(leakage_list) > 10 else ''}")

# ==============================================================================
# SAVE LEAKAGE RULES TO FILE (FOR TRAINING PIPELINE)
# ==============================================================================
import json

with open("leakage_rules.json", "w") as f:
    json.dump(leakage_report, f, indent=2)

print("\n💾 Saved leakage rules to: leakage_rules.json")

# ==============================================================================
# OPTIONAL: CREATE FEATURE SETS FOR EACH MODEL
# ==============================================================================
print("\n" + "="*70)
print("🔧 CREATING FEATURE SETS FOR EACH MODEL")
print("="*70)

all_numeric = df.select_dtypes(include=[np.number]).columns.tolist()

for target, leakage_list in leakage_report.items():
    # Remove leakage + target itself
    features = [col for col in all_numeric if col not in leakage_list and col != target]
    print(f"\n{target}: {len(features)} clean features available")
    
    # Save to file
    with open(f"features_{target}.txt", "w") as f:
        f.write("\n".join(features))

print("\n✅ Feature sets saved to individual files!")
print("\n🎉 YOU'RE NOW READY TO TRAIN MODELS WITHOUT LEAKAGE!")
```
